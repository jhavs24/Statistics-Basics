**Q.1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.**

Ans.

1. **Descriptive Statistics**

**Definition:**  Descriptive statistics are methods used to summarize, organize, and present data. They describe the main features of a dataset.

**Purpose:** To provide a clear picture of the collected data without making predictions or generalizations.

**Techniques Used:**
• Measures of Central Tendency: Mean, Median, Mode


• Measures of Dispersion: Range, Variance, Standard Deviation

• Data Visualization: Tables, Graphs, Charts

*Example:*

A teacher records the marks of 50 students in her class and calculates:

• Mean = 72

• Median = 70

• Mode = 68

• Standard Deviation = 5

These values only describe the performance of these 50 students.

 2. **Inferential Statistics**

**Definition:** Inferential statistics are methods used to draw conclusions, make predictions, or test hypotheses about a larger population based on a sample.

**Purpose:** To go beyond the available data and make inferences about the entire population.

**Techniques Used:**

• Hypothesis Testing (t-test, chi-square, ANOVA)

• Confidence Intervals

• Regression and Correlation

• Probability Distributions

*Example:*
A political analyst surveys 100 voters in a city and finds that 60% support Candidate A. Using inferential statistics, the analyst predicts that around 60% of all voters in the city will support Candidate A.

*Key Difference in One Line*

• Descriptive Statistics: Show what the data reveals.

• Inferential Statistics: Predict or conclude about a bigger group using sample data.


**Q.2: What is sampling in statistics? Explain the differences between random and stratified sampling.**

Ans.

**Sampling in Statistics:**

Sampling is the process of selecting a smaller group (called a sample) from a larger group (called the population) in order to study and analyze it6. Since it is often not possible to collect data from an entire population due to time, cost, or practical limits, researchers use samples to draw conclusions about the population.

*Example:* If a company wants to know the job satisfaction level of its 10,000 employees, it may collect responses from only 500 employees (sample) and use the results to represent the views of
the whole organization (population).

**Random Sampling**

**Definition:** Random sampling is a method in which every individual in the population has an equal chance of being selected. The selection is completely based on chance and is free from bias.

*Example:* If names of all students in a college are written on slips and 100 slips are drawn at random, then each student has the same chance of being selected.

**Stratified Sampling**

**Definition:** Stratified sampling is a method in which the population is first divided into subgroups or strata based on certain characteristics (such as age, gender, income, or education
level). Then random samples are taken from each stratum in proportion to its size.

*Example:* If a college has 60% male students and 40% female students, and we want a sample of 100, then stratified sampling will select 60males and 40 females randomly so that both groups are properly represented.

*Example*

Picking 100 students randomly from a list of names.
Picking 60 males and 40 females to match
population ratio.

**Q.3: Define mean, median, and mode. Explain why these measures of central tendency are important.**

Ans.

**Mean:** The mean (or average) is the value obtained by dividing the sum of all observations by the total number of observations.

**Formula:**
**Mean = Sum of all values / Number of values**

*Example:*

For marks 10, 20, 30, 40, 50:

Mean = 10+20+30+40+50 / 5 = 30

**Median:** The median is the middle value of a dataset when the values are arranged in ascending or descending order. If the number of observations is odd, the median is the middle value; if even, it is the average of the two middle values.

*Example:*

• For 10, 20, 30, 40, 50 → Median = 30

• For 10, 20, 30, 40 → Median = (20+30)/2 = 25

**Mode:** The mode is the observation that occurs most frequently in a dataset. A dataset can have one mode, more than one mode, or no mode at all.

*Example:* In 10, 20, 20, 30, 40, 50 → Mode = 20

**Importance of Mean, Median and Mode:**

1. Data Summarization: They condense a large set of data into a single value that represents the dataset.

2. Understanding Patterns: They help to identify the central or most typical value around which the data is distributed.

3. Decision-Making: Widely used in economics, business, education, and research for planning and forecasting.

4. Different Uses:

    a. Mean is best when data is evenly distributed.

    b. Median is useful when data has extreme values (outliers).

    c. Mode is most helpful in categorical data, such as finding the most popular product or preference.
In summary, mean, median, and mode are important measures of central tendency because they provide different ways of identifying the “center” of a dataset, making it easier to analyze, compare, and draw conclusions.

**Q.4: Explain skewness and kurtosis. What does a positive skew imply about the data?**

Ans.

**Skewness:** Skewness is a statistical measure that tells us about the symmetry or asymmetry of a distribution. It shows whether the data values are spread more towards the left or right side of the
mean.

• Symmetrical distribution: Skewness = 0 (example: Normal distribution).

• Positive skew (Right skew): Tail of the distribution is longer on the right side. Mean > Median > Mode.

• Negative skew (Left skew): Tail of the distribution is longer on the left side. Mean < Median < Mode.

**Kurtosis:** Kurtosis measures the peakedness or flatness of a distribution compared to a normal distribution. It shows how concentrated the data is around the mean.

• Mesokurtic: Kurtosis ≈ 3 → Normal distribution.

• Leptokurtic: Kurtosis > 3 → Distribution is more peaked (heavier tails).

• Platykurtic: Kurtosis < 3 → Distribution is flatter (lighter tails).

**What does a Positive Skew Imply?**

A positive skew (right-skewed distribution) implies that:

1. The data has a long tail on the right side.

2. Most of the values are concentrated on the lower side (left side) of the distribution.

3. The mean is greater than the median and mode.

4. Example: Income distribution in a population, where most people earn low or average income, but a few earn very high income, pulling the mean to the right.

In summary:

• Skewness shows the direction of data asymmetry.

• Kurtosis shows the sharpness or flatness of the data peak.

• Positive skew means data is stretched towards higher values with most observations on the lower side.

```python

# Q.6: Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python: list_x = [10, 20, 30, 40, 50] list_y = [15, 25, 35, 45, 60] (Include your Python code and output in the code box below.)

# Given datasets
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Step 1: Calculate the mean of each list
mean_x = sum(list_x) / len(list_x)
mean_y = sum(list_y) / len(list_y)

# Step 2: Calculate covariance
covariance = sum((x - mean_x)*(y - mean_y) for x, y in zip(list_x, list_y))
len(list_x)

# Step 3: Calculate standard deviations
std_x = (sum((x - mean_x)**2 for x in list_x) / len(list_x))**0.5
std_y = (sum((y - mean_y)**2 for y in list_y) / len(list_y))**0.5

# Step 4: Calculate correlation coefficient
correlation = covariance / (std_x * std_y)

# Display results
print("List X:", list_x)
print("List Y:", list_y)
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

```python
# Q.7: Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result: data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35] (Include your Python code and output in the code box below.)

# Import necessary library
import matplotlib.pyplot as plt

# Given dataset
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Draw the boxplot
plt.boxplot(data, vert=True, patch_artist=True)
plt.title("Boxplot of the Dataset")
plt.ylabel("Values")
plt.show()

# Identify outliers using the IQR method
data_sorted = sorted(data)
n = len(data_sorted)

# Find Q1 and Q3
Q1 = data_sorted[n // 4] # 25th percentile
Q3 = data_sorted[3 * n // 4] # 75th percentile
IQR = Q3 - Q1

# Outlier thresholds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

```python
# Q.8: You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.
# - Explain how you would use covariance and correlation to explore this relationship.
# - Write Python code to compute the correlation between the two lists: advertising_spend = [200, 250, 300, 400, 500] daily_sales = [2200, 2450,2750, 3200, 4000] (Include your Python code and output in the code box below.)

'''
Exploring Relationship between Advertising Spend and Daily Sales
Using Covariance and Correlation
1. Covariance:
Covariance measures how two variables change together. A positive covariance
indicates that when advertising spend increases, daily sales also tend to increase. A
negative covariance indicates that as one variable increases, the other tends to decrease.
However, covariance alone does not indicate the strength of the relationship because its
magnitude depends on the units of the variables.
2. Correlation:
Correlation (specifically, the Pearson correlation coefficient) measures both the strength
and direction of a linear relationship between two variables. It ranges from -1 to 1:
• 1 → Perfect positive correlation
• 0 → No correlation
• -1 → Perfect negative correlation
By computing correlation, we can determine how strongly advertising spend affects daily sales.
A high positive correlation would suggest that increasing ad spend is likely associated with
higher sales.
'''
# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Step 1: Calculate means
mean_ad = sum(advertising_spend) / len(advertising_spend)
mean_sales = sum(daily_sales) / len(daily_sales)

# Step 2: Calculate covariance
covariance = sum((a - mean_ad)*(s - mean_sales) for a, s in
zip(advertising_spend, daily_sales)) / len(advertising_spend)

# Step 3: Calculate standard deviations
std_ad = (sum((a - mean_ad)**2 for a in advertising_spend) /
len(advertising_spend))**0.5
std_sales = (sum((s - mean_sales)**2 for s in daily_sales) /
len(daily_sales))**0.5

# Step 4: Calculate correlation coefficient
correlation = covariance / (std_ad * std_sales)

# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```
